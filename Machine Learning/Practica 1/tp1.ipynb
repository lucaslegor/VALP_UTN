### ¿Qué es Pandas?  ¿Cuál es su relación con numpy?
Pandas es la libreria de Python mas utilizada para la manipulacion y el analisis de datos. Introduce dos estructuras principales que son las series (Columnas unidensionales) y los dataframes que son tablas bidimensionales (filas y columnas)
En machine learning, Pandas se encarga de la fase de preparacion:
  - Limpieza de datos
  - Feature engineering: Normalizacion de datos
  - Carga y exportacion: Lectura de BD's

La relacion que tiene con Numpy es que Pandas esta construido sobre Numpy. Pandas facilita la codificacion y entendimiento humano mientras que Numpy nos da la potencia matematica para calculo.

### ¿Por qué motivo un arreglo en numpy (ndarray) es más eficiente que una lista de Python?
Es mas eficiente ya que Numpy almacena los datos de forma contigua en la memoria y permite realizar operaciones matematicas en bloque sin usar bucles for.
