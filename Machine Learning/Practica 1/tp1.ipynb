### Descripción de un algoritmo: Regresión Lineal

La **Regresión Lineal** es un algoritmo de aprendizaje supervisado que permite predecir valores numéricos a partir de la relación entre una variable dependiente y una o varias variables independientes.

Su objetivo es encontrar una línea recta que se ajuste lo mejor posible a los datos, minimizando los errores entre los valores reales y los valores predichos.

La ecuación de la regresión lineal simple es:

$$
y = mx + b
$$

Donde:

- **y:** Valor que queremos predecir (variable dependiente).
- **x:** Variable que utilizamos para realizar la predicción (variable independiente).
- **m:** Pendiente de la recta, que indica cuánto cambia `y` cuando `x` aumenta una unidad.
- **b:** Intersección con el eje Y, es decir, el valor predicho cuando `x = 0`.

**Ejemplo práctico:**

En el dataset de salarios, podemos utilizar la regresión lineal para predecir el salario de una persona según sus años de experiencia laboral.

- **Variable independiente (X):** `YearsExperience`.
- **Variable dependiente (y):** `Salary`.

El algoritmo analiza los datos disponibles y encuentra una recta que representa la tendencia entre ambas variables. Una vez entrenado, podemos ingresar los años de experiencia de una persona y obtener una estimación de su salario.

## Imputacion de valores faltantes 
Por diversas razones, muchos conjuntos de datos del mundo real contienen valores faltantes, que a menudo se codifican como espacios en blanco, NaN (Not a Number), etc. Una estrategia básica para utilizar conjuntos de datos incompletos consiste en descartar filas y/o columnas completas que contengan valores faltantes. Sin embargo, esto implica la pérdida de datos que podrían ser valiosos (aunque estén incompletos). Una estrategia mejor es imputar los valores faltantes, es decir, inferirlos a partir de la parte conocida de los datos.

#### ¿Cuál es la diferencia entre imputación univariable (SimpleImputer) y multivariable (IterativeImputer)?
Un tipo de algoritmo de imputación es el univariante, que imputa valores en la i-ésima dimensión de característica utilizando únicamente los valores no faltantes de dicha dimensión (p. ej., SimpleImputer). Por el contrario, los algoritmos de imputación multivariante utilizan el conjunto completo de dimensiones de características disponibles para estimar los valores faltantes (p. ej., IterativeImputer).

```python
from sklearn.impute import MissingIndicator

mask = MissingIndicator(features="all").fit_transform(df)

print(pd.DataFrame(mask, columns=df.columns).sum())

'''
MissingIndicator() es una herramienta de Scikit-learn que identifica los valores faltantes.
features="all" indica que se deben incluir todas las columnas, incluso las que no tienen valores nulos.
fit_transform(df) genera una matriz de valores booleanos donde True indica un dato faltante.
pd.DataFrame(mask, columns=df.columns) convierte esa matriz en un DataFrame.
sum() cuenta cuántos valores faltantes hay por columna.
'''
```

Este ultimo codigo hace lo mismo que el codigo del inciso anterior solo que implementado scikit learn. Las salidas son iguales

#### Describa el propósito del siguiente fragmento de código. Señale las principales diferencias entre ambos mecanismos de imputación

```python
import pandas as pd
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import SimpleImputer, IterativeImputer
X = df[["city_development_index", "experience", "training_hours"]]
simple = SimpleImputer(strategy="median").fit_transform(X) # Estima los valores faltantes utilizando la mediana de cada columna.
iterative = IterativeImputer(random_state=0).fit_transform(X) # Estima los valores faltantes utilizando las relaciones entre las diferentes columnas del DataFrame.
rows, cols = np.where(X.isna().to_numpy()) # Identifica las posiciones de los valores faltantes.
imputed = pd.DataFrame({
"row": X.index[rows],
"column": X.columns[cols],
"original": X.to_numpy()[rows, cols],
"simple": simple[rows, cols],
"iterative": iterative[rows, cols],
}) # Crea un DataFrame que contiene la fila y columna de cada valor faltante, el valor original (NaN), el valor estimado por SimpleImputer y el valor estimado por IterativeImputer.
print(len(imputed))
print(imputed.head(12))
```

El fragmento de código tiene como objetivo identificar, reemplazar y comparar los valores faltantes (NaN) de tres columnas del DataFrame: city_development_index, experience y training_hours.
Para ello utiliza dos métodos de imputación de la librería Scikit-learn: SimpleImputer e IterativeImputer.
- **SimpleImputer (imputación simple)**: Reemplaza los valores faltantes de cada columna utilizando la mediana de los valores existentes de esa misma columna. 
    - Por ejemplo, si tenemos los siguientes años de experiencia: [2, 4, NaN, 8, 10].
    La mediana es 6, por lo que el valor faltante se reemplaza por 6. Este método no tiene en cuenta las relaciones entre las distintas variables.
- **IterativeImputer**: Estima los valores faltantes utilizando las relaciones entre las diferentes columnas del DataFrame.
    - Por ejemplo, para estimar un valor faltante de experience, puede utilizar la información disponible de city_development_index y training_hours.
