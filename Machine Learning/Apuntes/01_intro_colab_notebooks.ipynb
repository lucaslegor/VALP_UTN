> **¿Qué es un Notebook?**  
> Un notebook es un documento interactivo que combina **código ejecutable**, **texto explicativo**, **ecuaciones** e **imágenes** en una sola interfaz.

---

## 1. Anatomía de un Notebook

Un notebook está compuesto por **celdas** (cells). Hay dos tipos principales:

| Tipo | Descripción | Shortcut |
|------|-------------|----------|
| **Code** | Contiene código Python ejecutable | `Ctrl+Enter` para ejecutar |
| **Markdown** | Contiene texto con formato | Doble click para editar |

### Atajos de teclado esenciales

| Acción | Windows/Linux |
|--------|---------------|
| Ejecutar celda y avanzar | `Shift + Enter` |
| Ejecutar celda (quedarse) | `Ctrl + Enter` |
| Nueva celda arriba | `A` (en modo comando) |
| Nueva celda abajo | `B` (en modo comando) |
| Borrar celda | `D, D` (en modo comando) |
| Cambiar a Markdown | `M` (en modo comando) |
| Cambiar a Code | `Y` (en modo comando) |

## 7. Comandos mágicos (Magic Commands)

Jupyter tiene comandos especiales que empiezan con `%` (line magic) o `%%` (cell magic):

````python
# %timeit — mide el tiempo de ejecución
%timeit sum(range(10_000))
````

````python
# %%time — mide toda la celda
%%time
total = 0
for i in range(100_000):
    total += i
print(f"Suma: {total:,}")
````

````python
# %whos — muestra todas las variables en memoria
%whos
````
